```python
import time
import urllib.request

url = "http://127.0.0.1:8000/v1/models"

print("Đang chờ Qwen 8B load lên T4...")

for i in range(180):
    try:
        with urllib.request.urlopen(url, timeout=2) as response:
            if response.status == 200:
                print("✅ AI LOCAL ONLINE")
                print("✅ http://127.0.0.1:8000")
                break
    except Exception:
        pass

    time.sleep(2)

    if i % 10 == 0:
        print(f"Đã chờ {i * 2} giây...")
else:
    print("❌ AI chưa online.")
    print("\n===== LOG =====")
    !tail -n 60 /content/llama_server.log
```
